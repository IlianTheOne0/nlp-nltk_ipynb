```python
def preprocess_text(text):
    tokens = word_tokenize(text.lower())
    result = []

    for token in tokens:
        if token.isalpha() and token not in stop_words: result.append(token)
    return result

processed_corpus = [preprocess_text(doc) for doc in corpus]
print(f"An example of preprocessed sentence: {processed_corpus[2]}")
```

```python
size_1, win_1 = 20, 2
model_cbow_1 = Word2Vec(sentences=processed_corpus, vector_size=size_1, window=win_1, min_count=1, sg=0, epochs=100)
model_sg_1 = Word2Vec(sentences=processed_corpus, vector_size=size_1, window=win_1, min_count=1, sg=1, epochs=100)

size_2, win_2 = 100, 5
model_cbow_2 = Word2Vec(sentences=processed_corpus, vector_size=size_2, window=win_2, min_count=1, sg=0, epochs=100)
model_sg_2 = Word2Vec(sentences=processed_corpus, vector_size=size_2, window=win_2, min_count=1, sg=1, epochs=100)
```

```python
target_words = ["energy", "data", "structure"]

def evaluate_model(model, model_name, words):
    print(f"\n{model_name}")
    for word in words:
        try:
            similar = [w[0] for w in model.wv.most_similar(word, topn=3)]
            print(f"Similar to '{word}': {similar}")
        except KeyError: print(f"Word '{word}' is missing in the dictionary!")

print("Comparing:")
evaluate_model(model_cbow_1, f"CBOW (size={size_1}, window={win_1})", target_words)
evaluate_model(model_sg_1, f"Skip-gram (size={size_1}, window={win_1})", target_words)
evaluate_model(model_cbow_2, f"CBOW (size={size_2}, window={win_2})", target_words)
evaluate_model(model_sg_2, f"Skip-gram (size={size_2}, window={win_2})", target_words)
```

```python
word_pairs = [
    ("energy", "power"),
    ("data", "network"),
    ("structure", "design"),
    ("biological", "quantum")
]

print("\nCosine Similarity:")
for word_1, word_2 in word_pairs:
    if word_1 in model_cbow_1.wv and word_2 in model_cbow_1.wv:
        print(f"\nPair [{word_1} <-> {word_2}]:")
        print(f"  CBOW 1 (s={size_1}, w={win_1}): {model_cbow_1.wv.similarity(word_1, word_2):+.3f}")
        print(f"  SG 1   (s={size_1}, w={win_1}): {model_sg_1.wv.similarity(word_1, word_2):+.3f}")
        print(f"  CBOW 2 (s={size_2}, w={win_2}): {model_cbow_2.wv.similarity(word_1, word_2):+.3f}")
        print(f"  SG 2   (s={size_2}, w={win_2}): {model_sg_2.wv.similarity(word_1, word_2):+.3f}")
    else: print(f"\nPair [{word_1} <-> {word_2}] was skipped: one or both words are not in the dictionary!")
```

Результати на малому корпусі є некоректними із-за недостатньої кількості даних для навчання моделей (відбувається перенавчання; фіксуються випадкові співпадіння).

1. CBOW показує високі нереалістичні значення схожості для деяких пар слів, які не мають логічного зв’язку, а при збільшенні вектора до 100 розмірів косинусна схожість сягає +0.992 - із-за перенавчання на малому корпусі;
2. зв'язаними словами стають, наприклад, "energy", "like", "dna", тому що вони знаходяться сусідніми словами в 1-2 реченнях, а не через реальний зміст;
Искажение от гиперпараметров (Специфика Skip-gram): Увеличение vector_size до 100 и window до 5 на таком малом объеме создает избыточную размерность. Базовая логика моделей подтверждается, но стандартная обработка применяется с ошибками: Skip-gram не может собрать контекст и выдает хаотичные оценки (например, падение схожести structure и design до +0.026), так как ему не хватает примеров для предсказания контекста вокруг одного слова.
3.  збільшення розмірів векторів та контексту на малому обсязі корпусі свторює надмірну розмірність. Базова логіка моделей підтверджується, але стандартна обробка застосовується з помилками: Skip-gram не може зібрати контекст і видає хаотичні оцінки, тому що йому не вистачає прикладів для передбачення контексту навколо одного слова.
